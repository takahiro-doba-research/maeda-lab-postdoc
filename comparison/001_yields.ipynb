```python
file = "001_yields"
```

```python
rmse_atomic_pyridone = 6.180148
rmse_energy_pyridone = 6.197881
rmse_atomic_mpaa = 11.905044
rmse_energy_mpaa = 12.744916
```

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Patch

fig, ax = plt.subplots(figsize=(4.5, 4), dpi=300)

bar0 = ax.bar(0 - 0.175, rmse_atomic_pyridone, width=0.3, color="0.7")
bar1 = ax.bar(0 + 0.175, rmse_energy_pyridone, width=0.3, color="0.4")
bar2 = ax.bar(0.8 - 0.175, rmse_atomic_mpaa, width=0.3, color="0.7")
bar3 = ax.bar(0.8 + 0.175, rmse_energy_mpaa, width=0.3, color="0.4")

for bar in [bar0, bar1, bar2, bar3]:
    ax.bar_label(bar, fmt="%.1f", padding=3)

ax.set_title("Yield prediction")
ax.set_xticks([0, 0.8])
ax.set_xticklabels(["Pyridone", "MPAA"])
ax.set_ylabel("RMSE (test)")
ax.set_ylim(0, 14)

legend_elements = [
    Patch(facecolor="0.7", label="atomic descriptor"),
    Patch(facecolor="0.4", label="energy descriptor")
]
ax.legend(
    handles=legend_elements,
    loc="upper left"
)

plt.savefig(
    f"{file}_bar.png",
    transparent=True,
    bbox_inches="tight",
    pad_inches=0.1
)
plt.show()
```
