```python
import grrmlib as gl

gl.__version__
```

```python
eqs = gl.read_eq_list("td_c1_193_SCAFIR4_EQ_list.log")
```

# Carbopalladation complex

```python
eqs_filter = (
    eqs
    .filter(lambda eq: eq.get_distance(1, 25) > 2.25)
    .filter(lambda eq: eq.get_distance(13, 25) > 2.60)
    .filter(lambda eq: eq.get_distance(16, 25) < 2.10)
    .filter(lambda eq: eq.get_distance(7, 37) > 1.10)
    .filter(lambda eq: eq.get_distance(7, 35) < 1.10)
    .filter(lambda eq: eq.get_distance(1, 13) < 1.60)
    .filter(lambda eq: eq.get_distance(14, 25) > 1.60)
)
```

```python
eqs_filter
```

```python
eqs_filter_min = eqs_filter.smallest("scfenergy")
```

```python
eqs_filter_min.__dict__
```

```python
writer = gl.GaussianInputWriter()
writer.write(eqs_filter_min, f"EQ{eqs_filter_min.name}.com")
```
