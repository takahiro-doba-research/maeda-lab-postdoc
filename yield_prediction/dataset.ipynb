```python
import polars as pl

df_X = pl.read_csv("X.csv")
```

```python
df_X
```

```python
df_y = pl.read_csv("y.csv")
```

```python
df_y
```

```python
df = df_X.join(df_y, on=["backbone", "pyridone"], how="left", maintain_order="left")
```

```python
df
```

```python
for column in df.columns:
    if column.isnumeric() and column != "0":
        df = df.with_columns((pl.col(column) - pl.col("0")).alias(f"{column}_zero"))
```

```python
df = df.with_columns((pl.col("beta_av") / (100 - pl.col("beta_av"))).log().alias("beta_av_logit"))
```

```python
df = df.sort(["pyridone", "backbone"])
```

```python
df
```

```python
df.write_parquet("dataset.parquet")
```
