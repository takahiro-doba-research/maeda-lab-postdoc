```python
import polars as pl

df = pl.read_parquet("../../dataset.parquet")
```

```python
column_group = "pyridone"
column_fix = "backbone"
columns_feat = [column for column in df.columns if "zero" in column]
column_target = "beta_av_logit"
```

```python
import numpy as np
from sklearn.linear_model import Ridge
from sklearn.metrics import root_mean_squared_error
from sklearn.preprocessing import StandardScaler

def refit(
    df,
    column_group,
    column_fix,
    columns_feat,
    column_target
):
    df_preds = []
    df_coefs = []
    
    """
    Recursive feature elimination (RFE)
    """
    columns_rfe = columns_feat.copy()
    
    while len(columns_rfe) > 4:
        n_features = len(columns_rfe)
        print(f"n_features: {n_features}")
        
        X = df.select(columns_rfe).to_numpy()
        y = df.select(column_target).to_numpy()
        
        ss_X = StandardScaler()
        ss_y = StandardScaler()
        
        X_ = ss_X.fit_transform(X)
        y_ = ss_y.fit_transform(y)
        
        """
        Hyperparameter optimization
        """
        alpha = 100
        
        for i in range(1, -3, -1):
            rmses = {}
            
            for j in range(-9, 10):
                y_val_preds_ = []
                y_val_trues_ = []
                
                """
                Cross-validation
                """
                groups_val = df[column_group].unique(maintain_order=True)
                
                for group_val in groups_val:
                    df_train = df.filter(pl.col(column_group) != group_val)
                    df_val = df.filter(pl.col(column_group) == group_val)
                    
                    X_train = df_train.select(columns_rfe).to_numpy()
                    y_train = df_train.select(column_target).to_numpy()
                    X_val = df_val.select(columns_rfe).to_numpy()
                    y_val = df_val.select(column_target).to_numpy()
                    
                    X_train_ = ss_X.transform(X_train)
                    y_train_ = ss_y.transform(y_train)
                    X_val_ = ss_X.transform(X_val)
                    y_val_ = ss_y.transform(y_val)
                    
                    model = Ridge(alpha + j * 10**i)
                    model.fit(X_train_, y_train_)
                    
                    y_val_pred_ = model.predict(X_val_)
                    y_val_preds_.append(y_val_pred_)
                    y_val_trues_.append(y_val_)
                
                y_val_preds_ = np.concatenate(y_val_preds_)
                y_val_trues_ = np.concatenate(y_val_trues_)
                rmses[j] = root_mean_squared_error(y_val_trues_, y_val_preds_)
            
            j_min = min(rmses, key=rmses.get)
            alpha += j_min * 10**i
        
        """
        Refit with the best hyperparameter
        """
        model = Ridge(alpha)
        model.fit(X_, y_)
        
        """
        Store data
        """
        y_pred_ = model.predict(X_)
        y_pred = ss_y.inverse_transform(y_pred_)
        df_pred = pl.DataFrame(
            {
                "n_features": n_features,
                column_group: df[column_group],
                column_fix: df[column_fix],
                column_target: df[column_target],
                f"{column_target}_pred": y_pred.ravel()
            }
        )
        df_preds.append(df_pred)
        
        coef = model.coef_
        df_coef = (
            pl.DataFrame(coef, schema=columns_rfe)
            .with_columns(
                pl.lit(n_features).alias("n_features"),
                pl.lit(alpha).alias("alpha"),
            )
            .select(["n_features", "alpha"] + columns_rfe)
        )
        df_coefs.append(df_coef)
        
        """
        Feature elimination
        """
        column_eliminate = columns_rfe[np.argmin(np.abs(coef))]
        columns_rfe = [column for column in columns_rfe if column != column_eliminate]
    
    df_preds = pl.concat(df_preds, how="vertical_relaxed")
    df_coefs = pl.concat(df_coefs, how="diagonal_relaxed")
    return df_preds, df_coefs
```

```python
for ratio in np.arange(0.1, 1.1, 0.1):
    print(f"noise ratio: {ratio:.1f}")
    np.random.seed(42)  # The same random seed was used for each loop, while varying the noise level.
    target_std = df[column_target].std(ddof=0)
    noise = np.random.normal(0.0, target_std * ratio, df.height)
    df_noise = df.with_columns(pl.col(column_target) + pl.Series(noise))
    df_preds, df_coefs = refit(
        df_noise,
        column_group,
        column_fix,
        columns_feat,
        column_target
    )
    df_preds.write_parquet(f"{file}_preds_{ratio:.1f}.parquet")
    df_coefs.write_parquet(f"{file}_coefs_{ratio:.1f}.parquet")
```
