```python
file_read = "001_refit"
file_write = "005_feature_coefficient_correlation"
```

```python
import polars as pl

df_coefs = pl.read_parquet(f"{file_read}_coefs.parquet")
```

```python
columns_feat = [column for column in df_coefs.columns if "zero" in column]
```

```python
df_coefs_ = df_coefs.filter(pl.col("n_features") == 51)[columns_feat]
```

```python
df_coefs_
```

```python
data_coef = df_coefs_.to_numpy().ravel()
```

```python
data_coef
```

```python
df = pl.read_parquet("../../dataset.parquet")
```

```python
df_ = df[columns_feat].mean()
```

```python
df_
```

```python
data_feat = df_.to_numpy().ravel()
```

```python
data_feat
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(5, 5), dpi=300)

ax.scatter(data_coef, data_feat, s=5)

ax.set_xlabel("Coefficient")
ax.set_ylabel("Energy (Hartree)")

plt.savefig(
    f"{file_write}_scatter.png",
    transparent=True,
    bbox_inches="tight",
    pad_inches=0.1
)
plt.show()
```

```python
import numpy as np

np.corrcoef(data_coef, data_feat)
```
