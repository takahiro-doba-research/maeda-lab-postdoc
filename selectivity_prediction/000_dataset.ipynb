# 0. Energy descriptor for selectivity

```python
import polars as pl

df_X_beta = pl.read_csv("energy_descriptor_beta.csv").sort(["pyridone", "backbone"])
df_X_alpha = pl.read_csv("energy_descriptor_alpha.csv").sort(["pyridone", "backbone"])
```

```python
df_X_beta
```

```python
df_X_alpha
```

```python
columns_feat = df_X_alpha.drop(["arene", "alkene", "PG", "backbone", "pyridone", "0"]).columns
```

```python
df_energy = df_X_alpha[columns_feat] - df_X_beta[columns_feat]
```

```python
df_energy
```

```python
df_X = df_X_beta[["backbone", "pyridone"]].hstack(df_energy)
```

```python
df_X
```

# 1. Experimental selectivity

```python
df_y = pl.read_excel("combined.xlsx").sort(["pyridone", "backbone"])
```

```python
df_y
```

```python
df_y = df_y[["pyridone", "backbone", "alpha_av", "beta_av"]]
```

```python
df_y
```

```python
df_y = df_y.with_columns((pl.col("beta_av") / pl.col("alpha_av")).log().alias("beta_alpha"))
```

```python
df_y
```

# 2. Dataset for selectivity

```python
df = df_X.join(
    df_y[["backbone", "pyridone", "beta_alpha"]],
    on=["backbone", "pyridone"],
    how="left",
    maintain_order="left"
)
```

```python
df
```

```python
df.write_parquet("dataset_selectivity.parquet")
```
